Phase 2 – Data Understanding
Objectives
Load all datasets
Understand dataset structure
Inspect columns and data types
Identify missing values
Detect duplicates
Generate descriptive statistics

```python
import pandas as pd

customers = pd.read_csv("../Data/Raw/customers_dataset.csv")
orders = pd.read_csv("../Data/Raw/orders_dataset.csv")
order_items = pd.read_csv("../Data/Raw/order_items_dataset.csv")
payments = pd.read_csv("../Data/Raw/order_payments_dataset.csv")
products = pd.read_csv("../Data/Raw/products_dataset.csv")
reviews = pd.read_csv("../Data/Raw/order_reviews_dataset.csv")
sellers = pd.read_csv("../Data/Raw/sellers_dataset.csv")
category = pd.read_csv("../Data/Raw/product_category_name_translation.csv")
geolocation = pd.read_csv("../Data/Raw/geolocation_dataset.csv")

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Products": products,
    "Reviews": reviews,
    "Sellers": sellers,
    "Category": category,
    "Geolocation": geolocation
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

# Display Column Names

Understanding the available columns in each dataset helps identify primary keys,
foreign keys, and the information contained in each table.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    
    print("=" * 60)
    print(df.columns.tolist())
    print()
```

Preview Dataset
Display the first five rows of each dataset to understand its structure and sample records.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    print("=" * 60)
    display(df.head())
```

Missing Values
Identify missing values in each dataset before starting the data cleaning process.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    print("=" * 60)
    print(df.isnull().sum())
    print()
```

Duplicate Records
Identify duplicate rows that may affect the quality of analysis.

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

Statistical Summary
Generate summary statistics for numerical columns to understand their distribution.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    print("=" * 60)
    display(df.describe())
```
