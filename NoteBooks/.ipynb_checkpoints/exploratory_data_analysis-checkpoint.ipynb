Import Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import plotly.express as px

import warnings
warnings.filterwarnings("ignore")
```

Load Cleaned Dataset

```python
customers = pd.read_csv("../data/processed/customers_clean.csv")
orders = pd.read_csv("../data/processed/orders_clean.csv")
order_items = pd.read_csv("../data/processed/order_items_clean.csv")
order_payments = pd.read_csv("../data/processed/order_payments_clean.csv")
products = pd.read_csv("../data/processed/products_clean.csv")
order_reviews = pd.read_csv("../data/processed/order_reviews_clean.csv")
sellers = pd.read_csv("../data/processed/sellers_clean.csv")
geolocation = pd.read_csv("../data/processed/geolocation_clean.csv")
product_category_translation = pd.read_csv("../data/processed/product_category_translation_clean.csv")
```

Dataset Overview

```python
datasets = {
    "Customers": customers,
    "Geolocation": geolocation,
    "Order Items": order_items,
    "Order Payments": order_payments,
    "Order Reviews": order_reviews,
    "Orders": orders,
    "Products": products,
    "Product Category Translation": product_category_translation,
    "Sellers": sellers
}
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    print("=" * 60)

    print("Shape:", df.shape)

    display(df.head())

    display(df.describe(include="all"))
```

Order Status Distribution

```python
orders["order_status"].value_counts()
```

```python
status_counts = orders["order_status"].value_counts()

plt.figure(figsize=(8,5))
status_counts.plot(kind="bar")

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.show()
```

Monthly Orders Trend

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)
orders["YearMonth"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
monthly_orders = (
    orders.groupby("YearMonth")
    .size()
    .reset_index(name="Orders")
)

monthly_orders
plt.figure(figsize=(14,5))

plt.plot(
    monthly_orders["YearMonth"],
    monthly_orders["Orders"],
    marker="o"
)

plt.xticks(rotation=90)

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.show()
```

Orders by State

```python
orders_customers = orders.merge(
    customers,
    on="customer_id",
    how="left"
)
state_orders = (
    orders_customers["customer_state"]
    .value_counts()
)

state_orders
plt.figure(figsize=(12,5))

state_orders.plot(kind="bar")

plt.title("Orders by State")
plt.xlabel("State")
plt.ylabel("Orders")

plt.show()
```

Top 10 Customer Cities

```python
top_cities = (
    orders_customers["customer_city"]
    .value_counts()
    .head(10)
)

top_cities
plt.figure(figsize=(10,5))

top_cities.plot(kind="bar")

plt.title("Top 10 Customer Cities")
plt.xlabel("City")
plt.ylabel("Orders")

plt.show()
```

Total Revenue Generated
Calculate Total Revenue

Monthly Revenue Trend

```python
orders_payments = orders.merge(
    order_payments,
    on="order_id",
    how="inner"
)
orders_payments["order_purchase_timestamp"] = pd.to_datetime(
    orders_payments["order_purchase_timestamp"]
)

orders_payments["YearMonth"] = (
    orders_payments["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
monthly_revenue = (
    orders_payments
    .groupby("YearMonth")["payment_value"]
    .sum()
    .reset_index()
)

monthly_revenue
plt.figure(figsize=(14,5))

plt.plot(
    monthly_revenue["YearMonth"],
    monthly_revenue["payment_value"],
    marker="o"
)

plt.xticks(rotation=90)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (R$)")

plt.show()
```

Average Order Value (AOV)

```python
average_order_value = (
    orders_payments.groupby("order_id")["payment_value"]
    .sum()
    .mean()
)

print(f"Average Order Value: R$ {average_order_value:.2f}")
```

Top 10 Highest Revenue States

```python
revenue_state = (
    orders_customers
    .merge(order_payments, on="order_id")
    .groupby("customer_state")["payment_value"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

revenue_state
plt.figure(figsize=(10,5))

revenue_state.plot(kind="bar")

plt.title("Top 10 States by Revenue")
plt.xlabel("State")
plt.ylabel("Revenue (R$)")

plt.show()
```

Delivery Time Analysis

```python
orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days
orders["delivery_days"].describe()
```

Histogram

```python
plt.figure(figsize=(10,5))

plt.hist(
    orders["delivery_days"].dropna(),
    bins=30
)

plt.title("Delivery Time Distribution")
plt.xlabel("Days")
plt.ylabel("Orders")

plt.show()
```

Average Delivery Time by State

```python
delivery_state = (
    orders
    .merge(customers, on="customer_id")
    .groupby("customer_state")["delivery_days"]
    .mean()
    .sort_values()
)

delivery_state
plt.figure(figsize=(12,5))

delivery_state.plot(kind="bar")

plt.title("Average Delivery Time by State")
plt.xlabel("State")
plt.ylabel("Days")

plt.show()
```

Review Score Distribution

```python
review_avg = order_reviews["review_score"].mean()

print(f"Average Review Score: {review_avg:.2f}")
```

Average Review Score by Product Category

```python
review_products = (
    order_reviews
    .merge(order_items, on="order_id")
    .merge(products, on="product_id")
)
category_reviews = (
    review_products
    .groupby("product_category_name_english")["review_score"]
    .mean()
    .sort_values(ascending=False)
)

category_reviews.head(10)
plt.figure(figsize=(12,5))

category_reviews.head(10).plot(kind="bar")

plt.title("Top Rated Product Categories")
plt.xlabel("Category")
plt.ylabel("Average Review Score")

plt.show()
```

Top 10 Sellers by Revenue

```python
category_sales = (
    order_items
    .merge(products, on="product_id")
    .merge(order_payments, on="order_id")
)

category_revenue = (
    category_sales
    .groupby("product_category_name_english")["payment_value"]
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

category_revenue
```

```python
plt.figure(figsize=(14,6))

category_revenue.plot(kind="bar")

plt.title("Top Product Categories by Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue (R$)")

plt.show()
```

Correlation Analysis

```python
numerical_data = order_items[
    ["price", "freight_value"]
].copy()

numerical_data["payment_value"] = (
    order_payments["payment_value"]
    .reset_index(drop=True)
)
correlation = numerical_data.corr()

correlation
```

```python
plt.figure(figsize=(6,5))

plt.imshow(correlation, cmap="Blues")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Matrix")

plt.show()
```

# Business Insights

## Order Analysis
- Most orders were successfully delivered, indicating an efficient order fulfillment process.
- A small percentage of orders were canceled or unavailable, highlighting opportunities to improve inventory management and logistics.

## Customer Analysis
- Customer orders are concentrated in a few major states and cities, suggesting high-demand regions.
- These regions can be prioritized for targeted marketing campaigns and faster delivery services.

## Payment Analysis
- Credit card is the most preferred payment method among customers.
- Most customers choose to pay in a small number of installments, indicating a preference for manageable payment plans.

## Revenue Analysis
- Revenue shows noticeable monthly fluctuations, reflecting seasonal shopping behavior.
- Identifying peak sales periods can help optimize inventory planning and promotional campaigns.

## Delivery Analysis
- Average delivery times vary across different states.
- States with longer delivery times may benefit from improved logistics or additional warehouse locations.

## Product Analysis
- Certain product categories generate significantly higher revenue than others.
- High-performing categories should receive increased marketing focus and inventory investment.

## Seller Analysis
- A relatively small number of sellers contribute a large portion of total revenue.
- Supporting high-performing sellers while helping lower-performing sellers improve could increase overall marketplace performance.

## Customer Reviews
- Most review scores are positive, indicating strong customer satisfaction.
- Monitoring categories with lower ratings can help improve product quality and customer experience.

## Correlation Analysis
- Product price and payment values show a positive relationship.
- Freight charges have a weaker relationship with overall payment value, suggesting shipping costs contribute less to total order value.

# Business Recommendations

- Improve delivery performance in states with higher average delivery times.
- Increase marketing efforts in high-revenue states and cities.
- Promote top-selling product categories through personalized recommendations.
- Encourage customers to use preferred payment methods by offering discounts or cashback.
- Analyze reasons for canceled and unavailable orders to reduce operational losses.
- Provide incentives and support programs for sellers with lower sales performance.
- Improve customer satisfaction by addressing issues in low-rated product categories.
- Use seasonal demand trends to optimize inventory planning and promotional campaigns.

# Conclusion

This exploratory data analysis provided valuable insights into customer behavior, sales performance, payment preferences, delivery efficiency, seller performance, and product trends.

The findings from this analysis establish a strong foundation for feature engineering, dashboard development, and predictive machine learning models in the next phases of the project.
